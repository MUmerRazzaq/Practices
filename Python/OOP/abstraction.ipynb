# Abstract Class — Appliance

- Create an abstract class Appliance with:

    Abstract method turn_on().

    Abstract method turn_off().

- Create WashingMachine and Refrigerator classes that implement these methods.

- Try to instantiate Appliance directly (should fail).

```python
from abc import ABC, abstractmethod
class Appliance(ABC):
    @abstractmethod
    def turn_on(self):
        pass

    @abstractmethod
    def turn_off(self):
        pass
    
```

```python
class Refrigerator(Appliance):
    def turn_on(self):
        print("Refrigerator is now ON")
        return "Refrigerator is cooling"

    def turn_off(self):
        print("Refrigerator is now OFF")
        return "Refrigerator has stopped cooling"
```

```python
refrigerator = Refrigerator()
print(refrigerator.turn_on())
print(refrigerator.turn_off())
```

    Modify the previous example so the abstract method has some shared logic in the parent and the child classes call it using super().

```python
from abc import ABC, abstractmethod
class Appliance(ABC):

    def __init__(self, company_name):
        self.company_name = company_name

    def get_company_name(self):
        return self.company_name

    @abstractmethod
    def turn_on(self):
        pass

    @abstractmethod
    def turn_off(self):
        pass
    
```

```python
class Refrigerator(Appliance):
    def __init__(self, company_name):
        super().__init__(company_name)

    def turn_on(self):
        print("Refrigerator is now ON")
        return "Refrigerator is cooling"

    def turn_off(self):
        print("Refrigerator is now OFF")
        return "Refrigerator has stopped cooling"
```
