Polymorphism — Shape Area Calculator

    Create a base Shape class with a method area().

    Create subclasses Circle, Rectangle, and Triangle — each overrides area() with its own formula.

    Create a list of shapes and loop through them, calling area() on each.

```python
import math

class Shape:
    def area(self):
        raise NotImplementedError("Subclasses must implement this method")
    
    
```

```python
class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return math.pi * (self.radius ** 2)
```

```python
circle = Circle(5)
print(f"{circle.area():.2f}")
```

# Polymorphism — Payment Methods

- Create a base Payment class with process_payment().

    Create subclasses CreditCardPayment, PayPalPayment, and CryptoPayment.

    Loop through them and call the same method, showing different outputs.

```python
class Payment:
    def process(self):
        raise NotImplementedError("Subclasses must implement this method")
```

```python
class CreditCardPayment(Payment):
    def process(self):
        return "Processing credit card payment"
    
class PayPalPayment(Payment):
    def process(self):
        return "Processing PayPal payment"
    
class CryptoPayment(Payment):
    def process(self):
        return "Processing cryptocurrency payment"
```

```python
payments : Payment = [
    CreditCardPayment(),
    PayPalPayment(),
    CryptoPayment()]
```

```python
for payment in payments:
    print(payment.process())
```
