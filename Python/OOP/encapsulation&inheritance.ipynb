# Encapsulation + Inheritance

    Create a BankAccount base class with a private balance.

    Create SavingsAccount and CheckingAccount subclasses with different interest rates and withdrawal rules.

```python
class BankAccount:
    def __init__(self, balance=0):
        self.__balance = balance

    def get_balance(self):
        return self.__balance

    def deposit(self, amount):
        if amount < 0:
            raise ValueError("Deposit amount must be positive")
        self._update_balance(amount)
        return self.__balance

    def withdraw(self, amount):
        if 0 < amount <= self.__balance:
            self._update_balance(-amount)
            return self.__balance
        else:
            raise ValueError("Withdrawal amount must be positive and less than or equal to the balance")     
        
    def _update_balance(self, amount):
        self.__balance += amount
        return self.__balance
        
```

```python
class SavingsAccount(BankAccount):
    interest_rate = 0.05
    withdrawal_fee = 1
        

    def interest(self):
        interest_amount = self.get_balance() * self.interest_rate
        self._update_balance(interest_amount)
        return interest_amount


    def withdraw(self, amount):
        if amount < 0:
            raise ValueError("Withdrawal amount must be positive")
        if amount + self.withdrawal_fee > self.get_balance():
            raise ValueError("Insufficient funds for withdrawal and fee")
        self._update_balance(-amount - self.withdrawal_fee)
        return self.get_balance()
    
    

```

```python
class CheckingAccount(BankAccount):
    interest_rate = 0.02
    withdrawal_fee = 5

    def interest(self):
        interest_amount = self.get_balance() * self.interest_rate
        self._update_balance(interest_amount)
        return interest_amount

    def withdraw(self, amount):
        if amount < 0:
            raise ValueError("Withdrawal amount must be positive")
        if amount + self.withdrawal_fee > self.get_balance():
            raise ValueError("Insufficient funds for withdrawal and fee")
        self._update_balance(-amount - self.withdrawal_fee)
        return self.get_balance()
```

```python
acc = CheckingAccount(balance=1000)
```

```python
acc.get_balance()
```

```python
acc.withdraw(100)
```
