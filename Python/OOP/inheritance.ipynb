# Inheritance — Vehicles

    Create a base Vehicle class with attributes brand, model, and year.

    Create subclasses Car and Bike with extra attributes.

    Implement a method in Vehicle that prints basic info, and override it in Car and Bike to add extra details.

```python
class Vehicles:
    def __init__(self, brand, model, year):
        self.brand = brand
        self.model = model
        self.year = year

    def vehicle_info(self):
        return f"Year = {self.year}, Brand = {self.brand}, Model = {self.model}"
```

```python
class Car(Vehicles):
    def __init__(self, doors):
        super().__init__("Toyota", "Corolla", 2020)
        self.doors = doors
    def vehicle_info(self):
        base_info = super().vehicle_info()
        return f"{base_info}, Doors = {self.doors}"
    
```

```python
car = Car(doors=4)
car.vehicle_info()  # Output: Year = 2020, Brand Toyota, Model Corolla
```

```python
class Bike(Vehicles):
    def __init__(self, type_of_bike):
        super().__init__("Yamaha", "MT-15", 2021)
        self.type_of_bike = type_of_bike
    def vehicle_info(self):
        base_info = super().vehicle_info()
        return f"{base_info}, Type of Bike = {self.type_of_bike}"
```

```python
bike = Bike(type_of_bike="Sport")
bike.vehicle_info()  # Output: Year = 2021, Brand Yamaha, Model MT
```

# Inheritance — Employees

    Create a base Employee class.

    Create subclasses Manager, Developer, and Intern.

    Override a method calculate_salary() differently in each subclass.

```python
class Employee:
    def __init__(self, name, position, salary):
        self.name = name
        self.position = position
        self.salary = salary

    def calculate_salary(self):
        salary = self.salary * 1.1
        return  salary
    

```

```python
class Manager(Employee):
    def __init__(self, name, salary):
        super().__init__(name, "Manager", salary)

    def calculate_salary(self):
        return self.salary * 1.5
```

```python
manager = Manager("Alice", 100000)
print(manager.calculate_salary())  # Output: 150000.0
```

```python
class Intern(Employee):
    def __init__(self, name, salary):
        super().__init__(name, "Intern", salary)

    def calculate_salary(self):
        return self.salary * 0.5
```

```python
intern = Intern("Bob", 30000)
print(intern.calculate_salary()) 
```
