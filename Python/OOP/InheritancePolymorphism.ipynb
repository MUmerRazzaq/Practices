# Inheritance + Polymorphism

- Create a Transport base class with a method travel(distance).
- Subclasses Car, Bus, Bicycle override travel() differently.
- Loop through objects and call travel() polymorphically.

```python
class Transport:
    def travel(self, distance):
        raise NotImplementedError("Subclasses must implement this method")
```

```python
class Car:
    def travel(self, distance):
        return f"Car travels {distance} km by road."
    
class Bus:
    def travel(self, distance):
        return f"Bus travels {distance} km by road with many passengers."
    
class Bicycle:
    def travel(self, distance):
        return f"Bicycle travels {distance} km by cycling."
```

```python
transport = [Car(), Bus(), Bicycle()]
for vehicle in transport:
    print(vehicle.travel(10))
```
