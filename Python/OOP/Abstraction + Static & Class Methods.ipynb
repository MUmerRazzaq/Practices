# Abstraction + Static & Class Methods

- Create an abstract class Employee with an abstract method calculate_salary().
- Use a @classmethod to keep track of how many employees have been created.
-  Use a @staticmethod to validate salary input.



```python
from abc import ABC, abstractmethod

class Employee(ABC):
    emp_count = 0
    def __init__(self, name, salary):
        self.name = name
        self.validate_salary(salary)
        self.salary = salary
        Employee.emp_count += 1

    @staticmethod
    def validate_salary(salary):
        if salary < 0:
            raise ValueError("Salary cannot be negative")
        return True
    
    @abstractmethod
    def calculate_salary(self):
        pass

    @classmethod
    def get_employee_count(cls):
        return cls.emp_count
        
    
```

```python
class PartTimeEmployee(Employee):
    def __init__(self, name, salary, hours_worked):
        self.validate_salary(salary)
        self.hours_worked = hours_worked
        assert hours_worked > 0, "Hours worked cannot be negative"
        super().__init__(name, salary)

    def calculate_salary(self):
        return self.salary * self.hours_worked
```

```python
part_time_emp = PartTimeEmployee("Alice", 20, 15)
print(f"Part-time employee {part_time_emp.name} earns: {part_time_emp.calculate_salary()}")
print(f"Total employees: {Employee.get_employee_count()}")
```
