# 📚 Library Management System (OOP Practice Project)

## 🎯 Objective
Build a **Library Management System** in Python using **Object-Oriented Programming (OOP)** concepts:
- Encapsulation
- Inheritance
- Polymorphism
- Static & Class Methods

---

## 🛠 Features
1. **Add Library Items**
   - Books, Magazines, and EBooks
   - Track total items in the library

2. **Borrow & Return Items**
   - Reduce or increase available copies
   - Prevent borrowing when no copies are available

3. **Member System**
   - Members can borrow and return multiple items
   - Track which member has borrowed what

4. **Data Display**
   - Use **polymorphism** so each item type displays different details
   - Example:
     - Book → title, author, genre
     - Magazine → title, author, issue number
     - EBook → title, author, file size

---

## 📦 Class Structure

### 1. LibraryItem (Base Class)
**Attributes:**
- `title`
- `author`
- `copies`

**Methods:**
- `display_info()` → overridden by child classes (**Polymorphism**)
- `borrow_item()`
- `return_item()`
- `get_title()`, `get_author()`, `get_copies()`, `set_copies()`

**Class & Static Methods:**
- `show_total_items()` → tracks all items
- `validate_year()` → checks if a year is valid

---

**Extra Attribute:**

**Overrides:**
- `display_info()`

### 3. Magazine (Child Class of LibraryItem)

**Overrides:**
- `display_info()`

---

### 4. EBook (Child Class of Book)
**Extra Attribute:**

**Overrides:**
- `display_info()`

---

### 5. Member
**Attributes:**
- `name`

**Methods:**
- `borrow_item(item)`

- `view_borrowed()`

---

## 🔄 Class Interaction
- `Member` borrows `LibraryItem` objects (could be Book, Magazine, or EBook).
- Borrowing/returning updates copies via **encapsulation**.
- `display_info()` shows details differently for each type (**polymorphism**).
- `LibraryItem` keeps a total count of all items using a **class method**.
- A **static method** validates publication year before adding new items.

```python
from abc import ABC , abstractmethod
class LibraryItem(ABC):
    total_items = 0
    def __init__(self, title , author , copies ):
        self.__title = title
        self.__author = author
        self.__copies = copies
        LibraryItem.total_items +=1

    def get_title(self):
        return self.__title

    def get_author(self):
        return self.__author

    def get_copies(self):
        return self.__copies

    def set_copies(self,copies):
        self.validate_copies(copies)
        self.__copies = copies

    @abstractmethod   
    def display_info(self):
        pass

    @staticmethod
    def validate_copies(copies):
        if copies < 0 :
            raise ValueError ("Copies can not be negative")
        return True
        
    @classmethod
    def show_total_items(cls):
        return cls.total_items
    

    
```
