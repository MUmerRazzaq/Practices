```python
# 03_fahrenheit_to_celsius
def main():
    degrees_fahrenheit=float(input("Please Enter temperature in fahrenheit: "))
    degrees_celsius=(degrees_fahrenheit-32)*5.0/9.0
    print(f"Temperature:{degrees_fahrenheit}F = {degrees_celsius}C ")
    return
if __name__=="__main__":
    main()
```

```python
#05_triangle_perimeter
def main():
    length_side1=float(input("What is the length of side 1? "))
    length_side2=float(input("What is the length of side 2? "))
    length_side3=float(input("What is the length of side 3? "))
    perimeter=length_side1+length_side2+length_side3
    print(f"The perimeter of the triangle is {perimeter} ")
    return
if __name__=="__main__":
    main()
```

```python
# 1_dicesimulator
import random
dic_sides=6
def rool_dic():
    dic1=random.randint(1,dic_sides)
    dic2=random.randint(1,dic_sides)
    total=dic1+dic2
    print(f"Total of both dices = {total}")
    return

def main():
    dic1:int=10
    print(f"Dice 1 in main() is stared as {dic1}")
    rool_dic()
    rool_dic()
    rool_dic()
    print(f"Dice 1 in main() is End as {dic1}")
    return

if __name__=="__main__":
    main()
```
