```python
import csv

class Student:
    def __init__(self, name):
        self.name = name
        self.grades = {}  # Dictionary to hold subject-score pairs

    def add_grade(self, subject, score):
        """Add or update a grade for a specific subject."""
        self.grades[subject] = score

    def get_grades(self):
        """Retrieve the student's grades."""
        return self.grades

    def calculate_average(self):
        """Calculate the average of all subject scores."""
        if not self.grades:
            return 0  # Handle the case where there are no grades
        total_score = sum(self.grades.values())
        num_subjects = len(self.grades)
        return total_score / num_subjects

    def is_passing(self, passing_score=40):
        """Check if the student is passing all subjects based on a minimum passing score."""
        return all(score >= passing_score for score in self.grades.values())

    def to_csv_row(self):
        """Convert student data to a row format for CSV storage."""
        return [self.name] + [f"{subject}:{score}" for subject, score in self.grades.items()]

    @staticmethod
    def from_csv_row(row):
        """Create a Student object from a row of CSV data."""
        name = row[0]
        student = Student(name)
        for grade in row[1:]:
            subject, score = grade.split(':')
            student.add_grade(subject, int(score))
        return student

class PerformanceTracker:
    def __init__(self):
        self.students = {}

    def add_student(self, student):
        """Add a student to the tracker."""
        self.students[student.name] = student

    def save_to_csv(self, filename):
        """Save student data to a CSV file."""
        with open(filename, 'w', newline='') as csvfile:
            writer = csv.writer(csvfile)
            for student in self.students.values():
                writer.writerow(student.to_csv_row())

    def load_from_csv(self, filename):
        """Load student data from a CSV file."""
        try:
            with open(filename, 'r') as csvfile:
                reader = csv.reader(csvfile)
                for row in reader:
                    student = Student.from_csv_row(row)
                    self.add_student(student)
        except FileNotFoundError:
            print(f"File {filename} not found. Starting with an empty tracker.")

    def calculate_class_average(self):
        """Calculate the average score across all students."""
        if not self.students:
            return 0
        total_average = sum(student.calculate_average() for student in self.students.values())
        return total_average / len(self.students)

    def display_student_performance(self):
        """Print each student's average score and pass/fail status."""
        for student in self.students.values():
            average = student.calculate_average()
            status = "Pass" if student.is_passing() else "Fail"
            print(f"Student: {student.name}, Average Score: {average:.2f}, Status: {status}")

# Example usage
if __name__ == "__main__":
    tracker = PerformanceTracker()

    # Load existing data from CSV if available
    tracker.load_from_csv('student_data.csv')

    # Add new student data
    while True:
        name = input("Enter student name (or type 'done' to finish): ")
        if name.lower() == 'done':
            break
        scores_input = input("Enter student scores as 'subject:score' pairs separated by spaces: ")
        scores = [pair.split(':') for pair in scores_input.split()]
        student = Student(name)
        for subject, score in scores:
            student.add_grade(subject, int(score))
        tracker.add_student(student)

    # Save the updated data to CSV
    tracker.save_to_csv('student_data.csv')

    # Display student performance
    tracker.display_student_performance()
    print(f"Class Average Score: {tracker.calculate_class_average():.2f}")
```
